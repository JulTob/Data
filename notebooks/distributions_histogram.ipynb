# Histograms approaching a smooth distribution

Colab escape hatch for the live cell in the book chapter
[Distributions](https://jultob.github.io/Data/statistics/distributions/).

Try changing `sample_size` and `bin_count`, then re-run the cell.
As the sample grows, the histogram bars start hugging the smooth curve.

```python
import numpy as np
import matplotlib.pyplot as plt

sample_size = 200
bin_count   = 15

rng     = np.random.default_rng(42)
samples = rng.normal(0, 1, sample_size)

x_grid  = np.linspace(-4, 4, 400)
density = (1 / np.sqrt(2 * np.pi)) * np.exp(-0.5 * x_grid ** 2)

fig, ax = plt.subplots(figsize=(6, 4))
ax.hist(samples, bins=bin_count, density=True,
        color="tomato", alpha=0.7, label="Histogram")
ax.plot(x_grid, density, color="black", lw=2, label="Normal density")
ax.set_title(f"N = {sample_size} samples, {bin_count} bins")
ax.set_xlabel("x")
ax.set_ylabel("Density")
ax.legend()
plt.show()
```

# Poisson probability, quick check

Colab escape hatch for the live cell in the book chapter
[Poisson Distribution](https://jultob.github.io/Data/statistics/distributions/poisson.html).

Change `lam` (the average rate lambda) or `k` (the count you are asking
about), then re-run the cell.

```python
import math

lam = 4
k   = 2

probability = (lam ** k) * math.exp(-lam) / math.factorial(k)
print(f"P(X = {k}) with lambda = {lam}  ->  {probability:.4f}")
```

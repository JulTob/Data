## Basic r syntax

### Subtask:
Introduce fundamental R syntax, including variables, data types, and basic operations.


**Reasoning**:
Introduce fundamental R syntax by explaining variable assignment, data types, basic operations, and printing output as outlined in the instructions.



```python
# 1. Variable Assignment
# In R, you can assign values to variables using the '<-' operator.
my_variable <- 10
my_string <- "Hello, R!"

# 2. Data Types
# R has several fundamental data types:
# Numeric: Represents real numbers (can be integers or decimals).
numeric_var <- 3.14
# Integer: Represents whole numbers. To explicitly create an integer, append 'L'.
integer_var <- 100L
# Character: Represents strings.
character_var <- "This is a string."
# Logical: Represents boolean values (TRUE or FALSE).
logical_var <- TRUE

# 3. Basic Operations
# R supports standard arithmetic operations:
a <- 5
b <- 2
sum_result <- a + b
difference_result <- a - b
product_result <- a * b
division_result <- a / b
power_result <- a ^ b # Exponentiation

# 4. Printing Output
# Use the print() function to display the value of a variable or expression.
print(my_variable)
print(my_string)
print(numeric_var)
print(integer_var)
print(character_var)
print(logical_var)
print(sum_result)
print(difference_result)
print(product_result)
print(division_result)
print(power_result)
```
